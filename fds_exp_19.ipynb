```python

import numpy as np

# Create an array
a = np.array([10, 20, 30, 40, 50])
print("Array:", a)

# Array functions
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
print("Standard Deviation:", np.std(a))

# Array operations
print("Addition:", a + 5)
print("Multiplication:", a * 2)

# Reshape array
b = np.array([1, 2, 3, 4, 5, 6])
print("Reshaped Array:")
print(b.reshape(2, 3))

# Sort array
c = np.array([30, 10, 50, 20, 40])
print("Sorted Array:", np.sort(c))

# Find array dimensions and size
print("Dimensions:", a.ndim)
print("Size:", a.size)
```
