```python
import numpy as np
import pandas as pd

df=pd.read_csv('/content/Hotel_Dataset.xlsx - Hotel_Dataset.csv')
df
```

```python
df.duplicated()
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
len(df)
```

```python
index=np.array(list(range(0,len(df))))
df.set_index(index,inplace=True)
index
```

```python
df
```

```python
df.drop(['Age_Group.1'],axis=1,inplace=True)
df
```

```python
df.loc[df['CustomerID'] < 0, 'CustomerID'] = np.nan
df.loc[df['Bill'] < 0, 'Bill'] = np.nan
df.loc[df['EstimatedSalary'] < 0, 'EstimatedSalary'] = np.nan
df
```

```python
df.Age_Group.unique()
```

```python
df.Hotel.unique()
```

```python
df.FoodPreference.replace(['Vegetarian','veg'],'Veg',inplace=True)
df.FoodPreference.replace(['non-Veg'],'Non-Veg',inplace=True)
df
```

```python

est_salary_fill = round(df['EstimatedSalary'].mean()) if not pd.isna(df['EstimatedSalary'].mean()) else 0
no_of_pax_fill = round(df['NoOfPax'].median()) if not pd.isna(df['NoOfPax'].median()) else 0
rating_fill = round(df['Rating(1-5)'].median()) if not pd.isna(df['Rating(1-5)'].median()) else 0
bill_fill = round(df['Bill'].mean()) if not pd.isna(df['Bill'].mean()) else

df['EstimatedSalary'] = df['EstimatedSalary'].fillna(est_salary_fill)
df['NoOfPax'] = df['NoOfPax'].fillna(no_of_pax_fill)
df['Rating(1-5)'] = df['Rating(1-5)'].fillna(rating_fill)
df['Bill'] = df['Bill'].fillna(bill_fill)

df
```
