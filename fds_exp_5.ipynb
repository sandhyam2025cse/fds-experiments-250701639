```python
import numpy as np
array=np.random.randint(1,100,16) # randomly generate 16 numbers between 1 to 100
array
```

```python
def outDetection(array):
    sorted_array = sorted(array)
    Q1, Q3 = np.percentile(sorted_array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur

lr, ur = outDetection(array)
lr, ur
```

```python
import seaborn as sns
%matplotlib inline
sns.displot(array)
```

```python
new_array=array[(array>lr) & (array<ur)]
new_array
```

```python
sns.displot(new_array)
```
